# 🎬 Viral Clipper: free, from any device (iPad too)

Long video → viral vertical shorts with captions. Free and open source: [GitHub](https://github.com/officialtapesh/viral-clipper)

**How to use (2 steps):**
1. Tap **Runtime → Run all** (or the ▶ buttons below, top to bottom). Sign in with Google if asked, then tap **Run anyway**.
2. After about 2 minutes, a link like `https://xxxx.gradio.live` appears at the bottom. **Open it**, then upload your video (or paste a YouTube link), choose how many shorts, and tap **Make shorts**.

Keep this tab open while you use the app. The free GPU makes transcription fast.
YouTube sometimes blocks Google's servers; if a link fails, **upload the video file** instead.
